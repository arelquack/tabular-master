```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style visualisasi
sns.set_theme(style="whitegrid")

# Load data train dan test
train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')

print(f"Jumlah baris dan kolom train: {train_df.shape}")
print(f"Jumlah baris dan kolom test: {test_df.shape}")

train_df.head()
```

```python
# Cek ringkasan tipe data dan jumlah missing value
missing_info = pd.DataFrame({
    'Data Type': train_df.dtypes,
    'Missing Values': train_df.isnull().sum(),
    'Missing Percentage (%)': (train_df.isnull().sum() / len(train_df) * 100).round(2)
})

missing_info.sort_values(by='Missing Percentage (%)', ascending=False)
```

```python
# Cek apakah target seimbang (balanced) atau timpang (imbalanced)
fig, ax = plt.subplots(1, 2, figsize=(10,4))

# Persentase
train_df['Transported'].value_counts(normalize=True).plot(
    kind='pie', autopct='%1.1f%%', ax=ax[0], colors=['#66b3ff', '#ff9999'], startangle=90
)
ax[0].set_ylabel('')
ax[0].set_title('Distribusi Target (%)')

# Jumlah riil
sns.countplot(data=train_df, x='Transported', ax=ax[1], palette=['#66b3ff', '#ff9999'])
ax[1].set_title('Jumlah Riil Target')

plt.tight_layout()
plt.show()

print(train_df['Transported'].value_counts())
```
